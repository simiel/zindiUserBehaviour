```python
comment = pd.read_csv('monthly_grouped_data/grouped_comments.csv')
discussion = pd.read_csv('monthly_grouped_data/grouped_discussions.csv')
submission = pd.read_csv('monthly_grouped_data/submissions_grouped.csv')
user = pd.read_csv('monthly_grouped_data/grouped_users.csv')
train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')
```

```python
train.drop(['CompPart','Comment','Sub','Disc',],axis=1,inplace=True)
```

```python
comment.rename(columns={'UserID':'User_ID','CommentDate Year':'year','CommentDate Month':'month','CommentDate Day_of_week':'count'},inplace=True)
```

```python
comment.columns
```

```python
ultimate_merge = train.merge(comment,how='left',on=['User_ID','year','month'])
```

```python
ultimate_merge.fillna(0,inplace=True)
```

```python
ultimate_merge.head()
```

```python
discussion.columns
```

```python
discussion.rename(columns={'UserID':'User_ID','DiscDate Year':'year','DiscDate Month':'month','Count_of_DiscID':'countID'},inplace=True)
```

```python
ultimate_merge = ultimate_merge.merge(discussion[['User_ID', 'year', 'month','countID']],how='left',on=['User_ID','year','month'])
```

```python
ultimate_merge.fillna(0,inplace=True)
```

```python
ultimate_merge.head()
```

```python
ultimate_merge.countID.sum()
```

```python
ultimate_test = test.merge(comment,how='left',on=['User_ID','year','month'])
```

```python
ultimate_test = ultimate_test.merge(discussion[['User_ID', 'year', 'month','countID']],how='left',on=['User_ID','year','month'])
```

```python
ultimate_test.fillna(0,inplace=True)
```

```python
ultimate_test.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
ultimate_merge.columns
```

```python
ultimate_test.columns
```

```python
x=ultimate_merge[['month', 'year', 'count', 'countID']]
y=ultimate_merge['Target']
```

```python
model = RandomForestClassifier()
model.fit(x,y)
```

```python
model.score(x,y)
```

```python
t = ultimate_test[['month', 'year', 'count', 'countID']]
result = model.predict(t)
```

```python
result=pd.DataFrame(result)
```
