## DATA GROUPING

```python
#important libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Load files
train = pd.read_csv('data/Train.csv')
test = pd.read_csv('data/Test.csv')
samplesubmission = pd.read_csv('data/SampleSubmission.csv')
variable_definitions = pd.read_csv('data/VariableDefinitions.csv')
comments=pd.read_csv('data/Comments.csv')
competitionPartipation=pd.read_csv('data/CompetitionPartipation.csv')
competitions=pd.read_csv('data/Competitions.csv')
discussions=pd.read_csv('data/Discussions.csv')
submissions=pd.read_csv('data/Submissions.csv')
users=pd.read_csv('data/users.csv')
data_list = [train,test,samplesubmission,variable_definitions,comments,competitionPartipation,competitions,discussions,submissions,users]
```

```python
print(comments.head())
print(comments.columns)
```

```python
grouped_comments= comments.groupby(['UserID','CommentDate Year','CommentDate Month']).count().reset_index()
```

```python
grouped_comments.head()
```

```python
type(grouped_comments)
```

```python
discussions.head()
```

```python
grouped_discussions = discussions.groupby(['UserID','DiscDate Year','DiscDate Month','FeatureF']).count().reset_index()
```

```python
grouped_discussions.drop('DiscDate Day_of_week',axis=1,inplace=True)
```

```python
grouped_discussions.head()
```

```python
grouped_discussions = grouped_discussions.rename(columns={'DiscID':'Count_of_DiscID'})
```

```python
grouped_discussions.head()
```

```python
print(submissions.info())
print(submissions.head())
```

```python
submissions_grouped = submissions.groupby(['UserID','CompID','SubDate Year','SubDate Month','FeatureG']).count().reset_index()
```

```python
submissions_grouped.rename(columns={'SubDate Day_of_week':'submission_count'},inplace=True)
```

```python
submissions_grouped.head()
```

```python
print(users.info())
print(users.head())
```

```python
users.drop('UserDate Day_of_week',axis=1,inplace=True)
```

```python
users.head()
```

## SAVE GROUPED DATA TO CSV FILES

```python
grouped_comments.to_csv(r'grouped_comments.csv',index=False)
grouped_discussions.to_csv(r'grouped_discussions.csv',index=False)
submissions_grouped.to_csv(r'submissions_grouped.csv',index=False)
users.to_csv(r'grouped_users.csv',index=False)
```
